# Systems Report

### Systems Design

IronClad Technology, a defense company, requires an efficient and scalable visual identification system to verify the identity of personnel. The traditional methods involving ID cards and manual security checks are becoming untenable due to company growth and increased stress on security personnel. The proposed system aims to achieve high identification rates of employees, low false positive rates for non-employees, adaptability to different brightness conditions, and dynamic adjustment of personnel access permissions.

The system design for IronClad Technology's visual identification system consists of several key components and processes to ensure high identification accuracy, high security and relieve the need for security personnel. 

The image capture component is responsible for capturing images of personnel at entry points. These images are then preprocessed by resizing them to a consistent size, which helps ensure uniformity and facilitates efficient processing as input for the deep learning model. This preprocessing step is crucial for maintaining consistent image quality under varying conditions, thus enhancing the model's robustness.

At the core of the system is the deep learning model, which employs a convolutional neural network (CNN) architecture. Specifically, the system supports models such as ResNet-18 and ResNet-34, trained using SimCLR for self-supervised learning. The models are selected based on user input for image size and ResNet variant, ensuring flexibility and adaptability. The training process leverages a large dataset of labeled images, critical for correct personnel identification.

The system stores extracted embeddings in an efficient storage system, utilizing a KDTree for fast retrieval of nearest neighbors in the high-dimensional embedding space. This component is designed to handle millions of embeddings, ensuring optimized search performance and minimizing false positives by setting an appropriate threshold for similarity scores during nearest neighbor searches.

Dynamic adjustment of personnel access is a vital feature, allowing real-time addition and removal of personnel. This component generates new embeddings for new hires, adds them to the KDTree and embedding storage, and removes embeddings and associated images of departing personnel, ensuring the KDTree is updated to reflect the most current access permissions.

To maintain a comprehensive record of all access attempts and ensure system performance, the system includes robust logging and monitoring components. Access logs capture each identification attempt with details such as timestamp, probe image, and decision. This can allow for regular performance audits to ensure the system maintains high accuracy and low false positive rates, providing a reliable security solution for IronClad Technology.

This structured flow ensures each component operates seamlessly within the system, providing an efficient and scalable visual identification solution.

### Data, Data Pipelines and Model

#### Data 

The data used in the system falls into two categories: training data and entry point probe data.

* **Training Data:** This consists of a large collection of labeled images of personnel, captured under various lighting conditions and angles. These images are used to train the deep learning models, ensuring they can accurately extract features necessary for identification.

* **Entry Point Probe Data:** These are the images captured in real-time at the entry points. They are preprocessed and passed through the trained model to extract embeddings, which are then used for identification against the stored embeddings in the KDTree.

#### Data Pipelines

The data pipelines are designed to efficiently handle the flow of data from capture through preprocessing, model inference, and storage, ensuring seamless integration and performance.

**Training Pipeline:** Collects and preprocesses training data, trains the deep learning model using SimCLR for self-supervised learning, and saves the trained model to a specified path.

**Capture Pipeline:** Captures images of personnel at entry points and preprocesses them by resizing to a consistent size.

**Inference Pipeline:** Takes preprocessed images from the entry points, passes them through the trained model to extract embeddings, and uses the KDTree to find the nearest neighbors for identification.

**Access Control Pipeline:** Adds new employees by generating and storing new embeddings, removes departing employees by deleting their embeddings and associated images, and updates the KDTree to reflect current personnel access.


#### Data Files Created

The system generates and manages several types of files to ensure smooth operation and high performance. These include model files containing trained weights and architecture, embedding files storing high-dimensional feature embeddings for each personnel image, access logs documenting each identification attempt with details for auditing and monitoring.

#### Model

The deep learning model used in the system is a ResNet-based Convolutional Neural Network (CNN), which is a powerful architecture for image classification and feature extraction.  Residual networks (ResNets) are designed to ease the training of deep networks by addressing the vanishing gradient problem, allowing the network to learn more complex features. The system supports ResNet-18 and ResNet-34 models, which differ in the number of layers, allowing flexibility in choosing a model that balances complexity and performance.

These models are trained using SimCLR, a self-supervised learning framework that does not require labeled data for training. Instead, SimCLR leverages contrastive learning to train the model to distinguish between different images by maximizing the agreement between differently augmented views of the same image. This training approach enhances the model's ability to learn robust and generalizable features, which is crucial for accurate identification under varying conditions.

Once trained, the model extracts high-dimensional feature embeddings from input images. These embeddings capture essential characteristics of the images, which are then used for identifying personnel by comparing them against stored embeddings.

The system stores extracted embeddings in an efficient storage system, utilizing a KDTree for fast retrieval of nearest neighbors in the high-dimensional embedding space. This component is designed to handle millions of embeddings, ensuring optimized search performance and minimizing false positives by setting an appropriate threshold for similarity scores during nearest neighbor searches.

### Metrics

#### Offline Metrics

Offline metrics can be used to evaluate the performance of the model during the development phase using validation data. These metrics help in fine-tuning the model before deployment. The primary offline metrics used are precision at k, recall at k, average precision at k, and mean reciprocal rank at k.

* **Precision at k:** Measures the proportion of relevant items in the top k results. It indicates how many of the top k identified personnel are correctly identified employees.

* **Recall at k:** Measures the proportion of relevant items found in the top k results relative to the total number of relevant items. It indicates how well the system retrieves all relevant employees within the top k results.

* **Average Precision at k:** Averages the precision at each relevant item in the top k results, providing a single metric for the precision across different ranks.

* **Mean Reciprocal Rank at k:** The average of the reciprocal ranks of the first relevant item across multiple queries. It evaluates how quickly the first relevant item is retrieved in the top k results.

#### Online Metrics

Once deployed, the following metrics can be calculated to understand the system's performance under actual usage conditions.

* **Identification Rate:** The rate at which the system successfully identifies personnel at entry points. It is crucial for measuring the effectiveness of the system in a real-world setting.

* **False Positive Rate:** The rate at which the system incorrectly identifies non-personnel as personnel. Keeping this rate low is essential for security purposes.

* **Response Time:** The time taken by the system to process an image and make an identification decision. It measures the system's efficiency and user experience.

* **System Uptime:** The percentage of time the system is operational and available for use. High system uptime is critical for continuous security.

* **Access Log Accuracy:** The accuracy and completeness of access logs, ensuring that all identification attempts are recorded correctly for auditing purposes.

### Post-Deployment Policies

#### Monitoring and Maintenance Plan

Continuous monitoring and maintenance are critical to ensure the system's reliability and performance. The monitoring plan includes real-time tracking of system metrics such as identification rate, false positive rate, and system latency. Regular audits of the access logs help identify patterns and potential issues. Maintenance involves periodic updates to the model with new training data to improve accuracy and adaptability. Scheduled checks of hardware components, network stability, and software updates ensure the system remains robust and up-to-date.

#### Fault Mitigation Strategies

To ensure the system's resilience, several fault mitigation strategies can be implemented, including deploying redundant servers and databases to maintain operations during hardware failures, and implementing automatic failover mechanisms to switch to backup systems if primary systems encounter issues. The system also features robust error handling to gracefully manage unexpected situations and prevent crashes, alongside regular backups of critical data such as model files, embeddings, and access logs to prevent data loss. Additionally, automated alerts can be set up to detect unusual activity or performance degradation, enabling prompt resolution by the maintenance team.
