# Systems Report

### Systems Design

TextWave, as a leading information technology services company, is increasingly confronted with the challenge of managing an ever-growing volume of institution data. The sheer volume of data makes manual search methods untenable, and there is a pressing need for an advanced, automated system that can parse through billions of documents and return relevant, domain-specific information in human-interpretable natural language. To address these challenges, we have developed a scalable, adaptive information retrieval system that is tailored to meet the critical requirements of high relevancy, dynamic adaptability, massive data processing, and the ability to cite information sources accurately. The design of the information retrieval system is grounded in the need to efficiently handle the specific demands of large-scale institutional data. The system is composed of several key components, each designed to meet the critical requirements identified.

Achieving high relevancy in information retrieval, especially in domain-specific contexts, is a cornerstone of our system design. The system employs advanced natural language processing (NLP) techniques to ensure that the retrieval process is deeply rooted in the semantic understanding of the content. The core of this capability is the embedding-based search mechanism, which leverages pre-trained transformer models such as Sentence Transformers for embedding generation. These models are fine-tuned to understand the nuances of domain-specific language, thereby ensuring that the embeddings (vector representations of text) capture the intricacies of the information pertinent to the institution.

The system's KD-Tree-based search architecture plays a pivotal role in maintaining high relevancy. The KD-Tree is built upon embeddings extracted from the document corpus, allowing the system to efficiently identify the nearest neighbor embeddings that are most relevant to the query. By using multiple distance measures, such as Euclidean and Cosine similarity, the system can adapt the retrieval process to different types of queries, ensuring that the most contextually appropriate results are returned.

One of the critical challenges for TextWave is the dynamic nature of institutional data. New documents are constantly being generated, and the system must be capable of integrating this new information without compromising its retrieval performance. Our system is designed with adaptability in mind, allowing it to seamlessly incorporate new documents into the existing corpus.

The document management functionality within the system includes the ability to add and remove documents dynamically. When a new document is added, the system processes it to extract relevant sentence chunks, generate embeddings, and insert them into the KD-Tree index. This insertion process ensures that the system’s search capabilities are always up-to-date with the latest information. Similarly, when a document is removed, its corresponding embeddings are deleted from the KD-Tree, maintaining the integrity of the search index. This dynamic update mechanism is crucial for adapting to the evolving data landscape of TextWave.

Scalability is a non-negotiable requirement given the volume of data TextWave handles. The system is architected to process and retrieve information from billions of documents efficiently. This is achieved through a combination of distributed processing and efficient data structures.

The KD-Tree structure is pivotal for scaling the search across massive datasets. KD-Trees enable logarithmic time complexity for nearest neighbor searches, making it feasible to perform quick lookups even as the size of the corpus grows exponentially. Additionally, the system is designed to operate in a distributed environment, where document processing and embedding generation can be parallelized across multiple nodes. This distributed approach ensures that the system can handle large-scale data ingestion and search tasks without performance bottlenecks.

Citing the sources of retrieved information is essential, particularly in institutional contexts where the provenance of data is critical. The system is designed to not only retrieve relevant information but also to provide precise citations for each piece of information it returns.

Each document processed by the system is associated with metadata that includes the document’s filename, segment number, and the exact text that was embedded. When the system retrieves relevant sentence chunks in response to a query, it also returns this metadata, allowing users to trace the information back to its original source. This capability is integral to maintaining transparency and accountability in information retrieval, ensuring that users can always verify the origins of the data they are provided with.

### Data, Data Pipelines and Model

#### Data 

The data used in the system falls into two categories: document corpus data and the input question.

* **Document Corpus:** The document corpus consists of a vast collection of textual documents that are central to the system's operation. These documents represent the entirety of the knowledge base that the system searches and retrieves information from. The corpus includes a wide range of document types, such as institutional research papers, technical reports, policy documents, and historical archives. Each document within the corpus is typically unstructured, meaning it is presented in a raw text format without predefined labels or structure that would facilitate automated processing.

* **Input Question:** The input question is the data provided by the user seeking information from the system. This question is typically a natural language query, posed in natural English language. The input question is the trigger for the entire retrieval process, as it initiates the search through the document corpus to find the most relevant information. The system's ability to understand and semantically process the input question is critical to ensuring that the returned information is both accurate and contextually appropriate.


#### Data Pipelines

The data pipelines in the system are designed to transform raw textual data into a structured format that can be efficiently indexed, searched, and utilized for generating human-interpretable answers. The data pipeline consists of several stages, each responsible for a specific transformation or processing task.

**Data Ingestion:** The first stage of the data pipeline involves data ingestion, where raw textual documents are collected and brought into the system. These documents may be ingested from various sources, such as internal databases, file systems, or external APIs. Once ingested, the documents are stored in a centralized repository, specifically the storage/corpus directory, which serves as the corpus for further processing.

**Text Preprocessing:** After ingestion, the documents undergo preprocessing to prepare them for embedding generation. Preprocessing involves several steps:

- Text Normalization: This step includes cleaning the text by removing special characters, unnecessary whitespace, and normalizing case to ensure consistency.

- Sentence Chunking: The documents are split into smaller, manageable chunks, typically consisting of a few sentences each. This is crucial because embedding models operate at the sentence or chunk level, and smaller chunks allow for more precise embedding generation and retrieval.

- Tokenization: Although handled internally by the embedding model, tokenization is a key step in breaking down text into tokens (words, subwords, etc.) that the model can process.

**Embedding Generation:** Once the text is preprocessed, the next stage involves generating embeddings. An embedding is a dense vector representation of text that captures its semantic meaning in a multi-dimensional space. The system uses a pre-trained transformer model, such as Sentence Transformers, to convert each chunk of text into an embedding. These embeddings are crucial because they allow the system to perform semantic searches, where the meaning of the text is more important than the exact words used.

**Indexing:** After embeddings are generated, they are organized into a data structure that facilitates efficient search. The system uses a KD-Tree for this purpose. The KD-Tree is an efficient data structure that allows for quick nearest-neighbor searches in high-dimensional spaces, which is essential for retrieving the most relevant chunks of text when a query is made. The embeddings, along with their associated metadata (such as document name and chunk location), are inserted into the KD-Tree, creating an index that the system can query during information retrieval.

**Query Handling and Search:** When a user inputs a query, the query is processed to generate its embedding using the same embedding model. This query embedding is then used to search the KD-Tree for the nearest neighbor embeddings, representing the most semantically relevant chunks of text in the corpus. The system retrieves these chunks along with their metadata, which includes citation information and the exact location within the document.

**Answer Generation:** The final stage of the data pipeline involves answer generation. The system uses a BERT-based question-answering model to generate a coherent, human-interpretable response based on the retrieved text chunks. The question-answering model processes the question and the context provided by the retrieved chunks to generate an answer that directly addresses the user’s query. The system then returns this answer, along with citations to the original documents, ensuring that users can trace the information back to its source.


#### Model

The system employs two main models: the Embedding Model and the Question-Answering Model.

**Embedding Model:** The Embedding Model is based on pre-trained transformer architectures such as Sentence Transformers (e.g., all-MiniLM-L6-v2). This model is responsible for converting textual chunks into embeddings, which are vector representations of text in a high-dimensional space. These embeddings encapsulate the semantic meaning of the text, allowing the system to perform semantic searches rather than relying on simple keyword matching. The Embedding Model is critical for ensuring that the system can identify the most relevant chunks of text, even when the exact wording of the query differs from that in the documents.

The embeddings generated by this model are used as the foundation for the KD-Tree indexing, enabling efficient and scalable retrieval operations. The choice of the embedding model is crucial because it must be capable of capturing the nuances of the domain-specific language used within the institution.

**Question-Answering Model:** The Question-Answering Model is built on the BERT architecture, specifically a variant fine-tuned on the SQuAD dataset (bert-large-uncased-whole-word-masking-finetuned-squad). This model is designed to generate natural language answers to user queries based on the context provided by the retrieved text chunks. The model takes a question and a set of context chunks as input and outputs a concise, coherent answer that directly addresses the query.

The BERT-based model is particularly effective because of its ability to understand the relationship between the question and the context, making it well-suited for generating accurate and contextually appropriate answers. The model’s architecture allows it to focus on the most relevant parts of the context, ensuring that the answers generated are both relevant and informative.


### Metrics

#### Offline Metrics

Offline metrics are used during the development and testing phases to evaluate the performance of the system in a controlled environment. These metrics allow developers to refine the system before it is deployed in a live setting. The primary offline metrics used are precision at k, recall at k, average precision at k, and mean reciprocal rank at k.

* **Precision @ k:** Measures the proportion of relevant items in the top k results. This metric is crucial for ensuring that the system returns highly relevant information in response to user queries, which is critical for achieving high relevancy in domain-specific contexts.

* **Recall @ k:** Measures the proportion of relevant items found in the top k results relative to the total number of relevant items. Unlike Precision @ k, which focuses on the top-k results, Recall @ k assesses the system's ability to capture as many relevant chunks as possible.

* **Mean Reciprocal Rank:** Rank position of the first relevant result returned by the system. It is calculated as the reciprocal of the rank of the first relevant document, averaged across multiple queries.MRR provides insight into how quickly the system retrieves a relevant result in the ranking order. A higher MRR indicates that relevant documents are appearing closer to the top of the search results, which is desirable for user satisfaction and efficiency in information retrieval.

* **Exact Match (EM):** Percentage of cases where the system's generated answer exactly matches the ground truth answer. EM is used to assess the accuracy of the generated answers in scenarios where the correctness of the response is binary (correct or incorrect. It is crucial for evaluating how often the system provides the exact correct answer to user queries.

* **Transformer Match:** Semantic similarity between the generated answer and the ground truth answer using transformer-based models. Unlike Exact Match, which looks for an exact string match, Transformer Match assesses how semantically close the generated answer is to the correct answer by using a model like BERT to compute similarity scores between the two texts. Transformer Match provides a more nuanced evaluation of the system's performance by capturing the semantic meaning of the generated answers. This is particularly useful in cases where the exact wording of the answer may differ, but the meaning is equivalent or very similar to the ground truth. A high Transformer Match score indicates that the generated answer conveys the same or very similar meaning as the ground truth, even if the wording is different.

* **F1 Score:** mean of precision and recall, providing a single metric that balances the trade-off between these two measures. The F1 score gives a more balanced view of the system’s performance, especially in situations where a compromise between precision and recall is necessary. It is a robust metric for evaluating the overall effectiveness of the information retrieval and question-answering components.

#### Online Metrics

Once deployed, the following metrics can be calculated to understand the system's performance under actual usage conditions.

* **User Satisfaction Rating:** User satisfaction is often measured through direct feedback or surveys where users rate their experience with the system. This metric is subjective but provides valuable insights into the perceived quality of the system’s responses. This is often helpful when the response is subjective.

* **Query Success Rate:**  Percentage of queries that are successfully processed and return valid results. This metric tracks the effectiveness of the system in handling user requests without errors or failures. A high Query Success Rate indicates that the system is robust and capable of handling diverse queries effectively. Monitoring this metric helps in identifying any issues that might cause queries to fail, such as data inconsistencies, system errors, or unhandled query types.

* **Latency:** Time taken by the system to process a query and return results. This metric is crucial for maintaining a responsive user experience, particularly when dealing with large document corpuses, containing potentially billions of documents. Ensuring low latency is critical for user satisfaction, especially in environments where quick access to information is essential. Monitoring response time helps identify and address performance bottlenecks that could slow down the system.

### Post-Deployment Policies

#### Monitoring and Maintenance Plan

Post-deployment, the information retrieval system will be subject to continuous monitoring to ensure optimal performance, reliability, and user satisfaction. This plan involves the implementation of automated monitoring tools that track key metrics such as system uptime, latency, query success rate, error rate, and data freshness. Regular log reviews and performance audits will be conducted to detect any anomalies or degradations in service. The system will also be maintained through scheduled updates, which include patching security vulnerabilities, optimizing data pipelines, and upgrading models to incorporate the latest advancements in natural language processing. Maintenance windows will be communicated in advance to minimize disruption, and real-time dashboards will be employed to provide ongoing visibility into system health. Additionally, a dedicated support team will be on standby to address any issues that arise, ensuring that the system remains robust and responsive to user needs.

#### Fault Mitigation Strategies

To ensure the system's resilience, a comprehensive fault mitigation strategy will be implemented to handle potential failures and minimize their impact on operations. This strategy includes the deployment of redundant infrastructure to prevent single points of failure, such as mirrored servers and databases that can take over in the event of an outage. Automated failover mechanisms will be configured to ensure that service continuity is maintained even during unexpected disruptions. In the event of a fault, real-time alerts will be triggered to notify the technical team, who will follow predefined incident response protocols to diagnose and resolve the issue quickly. The system will also include robust error handling mechanisms to manage errors without affecting the user experience, such as retry logic for transient failures and fallback procedures for critical failures. Regular disaster recovery drills will be conducted to ensure that the team is prepared to respond effectively to various fault scenarios, minimizing downtime and data loss.
